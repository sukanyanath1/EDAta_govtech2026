```python
import pandas as pd

df = pd.read_csv("../../data/economics/OECD-WTO_BATIS_BPM6_December2025_bulk.csv")
```

```python
df_ch = df[(df["Reporter"] == "CH") | (df["Partner"] == "CH")]
```

```python
df_ch = df_ch[df_ch["type_Partner"] == "c"] # only partner country, not world or group
# remove type_Partner column as it is no longer needed
df_ch.drop(columns=["type_Partner"], inplace=True)

df_ch = df_ch[df_ch["type_Reporter"] == "c"] # only reporter country, not world or group
# remove type_Reporter column as it is no longer needed
df_ch.drop(columns=["type_Reporter"], inplace=True)
```

```python
item_codes_to_keep = [
    "S", "SA", "SB", "SC", "SD", "SE", "SF", "SG", "SH",
    "SI", "SJ", "SK", "SL"
]

df_ch = df_ch[df_ch["Item_code"].isin(item_codes_to_keep)].copy()
```

```python
df_ch.to_csv("../../data/economics/OECD-WTO_BATIS_BPM6_December2025_CH.csv", index=False)
```
